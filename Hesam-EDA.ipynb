# important variables and their distribtutions:
In the EDA phase (not feature selection yet), "important" means variables that are most likely to help us understand the data and predict y. Here's how to identify them:

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Distribution of target
sns.countplot(x='y', data=df)
plt.title('Distribution of Target Variable y')
plt.show()

# Optional: print percentages
print(df['y'].value_counts(normalize=True) * 100)
```
